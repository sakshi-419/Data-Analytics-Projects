```python
## "What percentage of Netflix content is family-friendly versus adult-oriented?"
family_ratings = ['TV-Y', 'TV-Y7', 'TV-G', 'G', 'PG', 'TV-PG']

df['audience_type'] = df['rating'].apply(
    lambda x: 'Family-Friendly' if x in family_ratings else 'Adult-Oriented'
)
```

```python
audience_percent = (
    df['audience_type']
    .value_counts(normalize=True) * 100
)

print(audience_percent)
```
