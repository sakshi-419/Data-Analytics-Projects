```python
df = df.dropna(subset=['CustomerID'])  ## the customer which has no customerid remove
df
```

```python
df = df.drop_duplicates()  ## revome the duplicate rows
df
```

```python
## Remove  cancelled orders
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
df
```

```python
## Remove negative quantities
df = df[df['Quantity'] > 0]
df
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
df
```

```python
#Which country generates the highest revenue?
country_sales = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False)

country_sales.head(10)
```

```python
country_sales.head(10).plot(kind='bar')
plt.title('Top 10 Countries by Revenue')
plt.show()
```

```python
# What are the top-selling products?


```

```python
##What are the top-selling products?
top_products = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False)
top_products.head(10)
```

```python
## Who are the top customers?
top_customers = df.groupby('CustomerID')['Revenue'].sum().sort_values(ascending=False)
top_customers.head(10)
```

```python
## Which month generates the most revenue?
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

df['Month'] = df['InvoiceDate'].dt.month_name()

monthly_sales = df.groupby('Month')['Revenue'].sum()

monthly_sales
```

```python
## What percentage of sales comes from the top 10 customers?
top10 = top_customers.head(10).sum()

total = df['Revenue'].sum()

(top10/total)*100
```

```python
##What is the average order value?
order_value = df.groupby('InvoiceNo')['Revenue'].sum()

order_value.mean()
```
